# 风机控制工作区的简化示例

这个示例用归一化变量画出两类典型控制关系：额定以下的发电机转矩控制，以及额定以上的功率限制与变桨。

曲线只用于说明控制结构，不对应具体机型或控制器参数。

```python
import numpy as np
import matplotlib.pyplot as plt

omega = np.linspace(0.2, 1.2, 500)
omega_rated = 1.0
k_opt = 1.0

torque = np.where(
    omega <= omega_rated,
    k_opt * omega**2,
    1.0 / omega
)

power = torque * omega

pitch_deg = np.where(
    omega <= omega_rated,
    0.0,
    20.0 * (omega - omega_rated) / (1.2 - omega_rated)
)

print(f"rated normalized speed = {omega_rated:.1f}")
print(f"rated normalized power = {np.interp(omega_rated, omega, power):.3f}")
```

额定以下采用

$$
\tau_g \propto \omega_g^2,
$$

可使机组在简化条件下沿接近最佳叶尖速比的工作线运行。

达到额定后，示例把电功率限制在归一化值1附近，因此转矩近似按

$$
\tau_g=\frac{P_{rated}}{\omega_g}
$$

下降。同时让桨距角逐步增加，用来表示额定以上通过变桨削减气动输入。

```python
plt.figure(figsize=(8, 4.5))
plt.plot(omega, torque, label="Generator torque")
plt.plot(omega, power, label="Electrical power")
plt.axvline(omega_rated, linestyle="--", label="Rated speed")
plt.xlabel("Normalized generator speed")
plt.ylabel("Normalized value")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4.5))
plt.plot(omega, pitch_deg)
plt.axvline(omega_rated, linestyle="--")
plt.xlabel("Normalized generator speed")
plt.ylabel("Illustrative pitch angle (deg)")
plt.tight_layout()
plt.show()
```

## 读图时保留的边界

真实风机控制会包含启动、停机、转矩和桨距限幅、执行器动态、滤波、增益调度、载荷缓解和故障逻辑。不同机型的分区边界和控制律也会不同。

这个示例只保留“额定以下追求高能量捕获、额定以上限制功率和转速”这一控制骨架。
